### 9. Username generator

Given a first name (string) and a birth year (int), build a lowercase, space-free username: `"Ethan"` + `1973` → `"ethan1973"`.

```python
first = "Ethan"
year = 1973

print(first.lower() + str(year))
```

**Extend:** wrap it in a function `make_username(first, year)` that also strips spaces from multi-word names ("Ethan Weed" → "ethanweed1973"), then test it on your own name.

```python
def make_username(first, year):
    print(first.lower() + str(year))

make_username('Christos', 2002)
```
